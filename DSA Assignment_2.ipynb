## Part I: Binary Tree Diagram

```python
from collections import deque

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Construct tree
A = Node('A')
B = Node('B')
C = Node('C')
D = Node('D')
E = Node('E')
F = Node('F')

A.left = B
A.right = C
B.left = D
B.right = E
C.right = F
```

#### Traversal Functions

```python
def preorder(node):
    if node:
        print(node.data, end=" ")
        preorder(node.left)
        preorder(node.right)


def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=" ")
        inorder(node.right)


def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=" ")
```

#### Level Order Traversal

```python
def level_order(root):
    if not root:
        return

    queue = deque([root])

    while queue:
        node = queue.popleft()
        print(node.data, end=" ")

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)
```

#### Run Traversals

```python
print("Pre-order:")
preorder(A)

print("\n\nIn-order:")
inorder(A)

print("\n\nPost-order:")
postorder(A)

print("\n\nLevel-order:")
level_order(A)
```

## Part II: Binary Search Tree

Values: 50, 30, 70, 20, 40, 60, 80

```python
class BSTNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


def insert(root, value):

    if root is None:
        return BSTNode(value)

    if value < root.value:
        root.left = insert(root.left, value)
    else:
        root.right = insert(root.right, value)

    return root
```

#### Search Function

```python
def search(root, key):

    if root is None:
        return False

    if root.value == key:
        return True

    if key < root.value:
        return search(root.left, key)

    return search(root.right, key)
```

#### Build BST

```python
values = [50, 30, 70, 20, 40, 60, 80]

root = None

for v in values:
    root = insert(root, v)
```

#### Search Values

```python
print("40 found:", search(root, 40))
print("90 found:", search(root, 90))
```

#### Report

| Search Value | Result    |
| ------------ | --------- |
| 40           | Found     |
| 90           | Not Found |

40 is found because it exists in the BST as the right child of 30.
90 is not found because no node with value 90 exists in the tree.

```python
class BSTNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


def insert(root, value):

    if root is None:
        return BSTNode(value)

    if value < root.value:
        root.left = insert(root.left, value)
    else:
        root.right = insert(root.right, value)

    return root
```

#### Height Function

```python
def height(node):

    if node is None:
        return -1

    return 1 + max(
        height(node.left),
        height(node.right)
    )
```

#### Build Tree

```python
root2 = None

for value in [10, 20, 30, 40, 50]:
    root2 = insert(root2, value)

print("Height =", height(root2))
```

## Part VI: Bioinformatics Application

#### Build the Tree

```python
from collections import deque

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Create nodes
ancestor = Node("Ancestor")

mammals = Node("Mammals")
birds = Node("Birds")

human = Node("Human")
cow = Node("Cow")

# Build phylogenetic tree
ancestor.left = mammals
ancestor.right = birds

mammals.left = human
mammals.right = cow
```

#### Traversal Functions

```python
def preorder(node):

    if node:
        print(node.data, end=" ")
        preorder(node.left)
        preorder(node.right)


def postorder(node):

    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.data, end=" ")


def level_order(root):

    if not root:
        return

    queue = deque([root])

    while queue:

        current = queue.popleft()
        print(current.data, end=" ")

        if current.left:
            queue.append(current.left)

        if current.right:
            queue.append(current.right)
```

#### Perform Traversals

```python
print("Pre-order Traversal:")
preorder(ancestor)

print("\n")

print("Post-order Traversal:")
postorder(ancestor)

print("\n")

print("Level-order Traversal:")
level_order(ancestor)
```

#### Height Function

```python
def height(node):

    if node is None:
        return -1

    return 1 + max(
        height(node.left),
        height(node.right)
    )
```

#### Determine Tree Properties

```python
print("Root Node:")
print("Ancestor")

print("\nInternal Nodes:")
print("Ancestor, Mammals")

print("\nLeaf Nodes:")
print("Human, Cow, Birds")

print("\nHeight:")
print(height(ancestor))
```

#### Interpretation:

Interpretation:

This phylogenetic tree represents evolutionary relationships among organisms.

• Ancestor is the common ancestor of all species in the tree.

• Mammals and Birds are major evolutionary groups.

• Human and Cow share a more recent common ancestor (Mammals) than they do with Birds.

• The tree height of 2 indicates that the longest evolutionary path from the root to a leaf contains two edges.

Tree traversals are useful in bioinformatics for exploring phylogenetic relationships, evolutionary history, and hierarchical biological classifications.
