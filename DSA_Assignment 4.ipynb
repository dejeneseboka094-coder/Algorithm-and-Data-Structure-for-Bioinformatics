### Question 1: Hashing Fundamentals

### (a) Define the following terms

1. Hash Table

A hash table is a data structure used to store and retrieve data efficiently. It uses a hash function to convert a key into an index of an array, where the corresponding data is stored.

A hash table generally consists of:

A collection of keys
A hash function
An array containing the table locations
A collision-resolution mechanism

The expected running time for insertion, searching, and deletion in a hash table is O(1).

2. Hash Function

A hash function is a function that maps a key from a large set of possible keys into a smaller range of hash-table indices.

It can be represented as:

h:U→{0,1,2,…,m−1}

where:

U = universe of possible keys
m = number of locations in the hash table
h(k) = hash value or index of key k

A simple example is the division method:

h(k)=kmodm

For example, if m=10:

h(1087)=1087mod10=7

Therefore, key 1087 is stored at index 7.

3. Hash Collision

A hash collision occurs when two different keys produce the same hash-table index.

Mathematically:

         h(k1) = h(k2)
For example, if:
     h(k)=kmod7
     15mod7=1 and 22mod7=1
Therefore, both 15 and 22 map to index 1, producing a collision.


### (b) Explain five characteristics of a good hash function

A good hash function should have the following characteristics:

1. Deterministic

A hash function must always produce the same hash value for the same key.

For example, if:

   h(25)=3

then every time the key 25 is hashed, the result should be 3.

2. Computationally efficient

The hash function should be fast to calculate because hashing may be performed many times during insertion, searching, and deletion.

A computationally efficient hash function helps maintain the expected O(1) performance of hash-table operations.

3. Uniform distribution

A good hash function should distribute keys uniformly across the available table locations.

This prevents too many keys from being concentrated in a small number of locations.

4. Minimize collisions

A good hash function should minimize the number of different keys that produce the same index.

Fewer collisions generally result in better hash-table performance.

5. Use information from the entire key and avoid clustering

A good hash function should use the relevant information contained in the key rather than relying only on a small portion of it.

It should also avoid clustering, where many keys become concentrated in nearby table locations.

the main characteristics include being deterministic, computationally efficient, distributing keys uniformly, minimizing collisions, using information from the entire key, and avoiding clustering.

### (c) Why are collisions unavoidable in practical hash tables?

Collisions are unavoidable because the number of possible keys is normally much larger than the number of available locations in the hash table.

For example, suppose a hash table has only 10 locations:

0,1,2,…,9

but there are thousands or millions of possible keys.

Since many different keys must be mapped into only 10 locations, some different keys will inevitably receive the same index.

Therefore:

Collisions are unavoidable because the number of possible keys is normally much larger than the number of table locations.

For this reason, hash tables require collision-resolution techniques such as separate chaining, linear probing, quadratic probing, and double hashing.

## Question 2: Collision Resolution

When two different keys produce the same hash-table index, a collision occurs. A hash table therefore needs a method for resolving collisions.

The four collision-resolution techniques considered in this assignment are:

 ##### Separate chaining
 ##### Linear probing
 ##### Quadratic probing
 ##### Double hashing

#### 1 Separate Chaining

Separate chaining stores multiple elements that hash to the same location in a secondary data structure, traditionally a linked list.

For example:
  Index 0 → NULL
  Index 1 → 15 → 29 → NULL
  Index 2 → NULL
  Index 3 → 10 → 17 → 24 → NULL
  Index 4 → NULL

If several keys produce the same hash value, they are stored together in the chain at that index.

##### Advantages
    
Simple to implement

Multiple elements can be stored at the same index

The table does not need to find another empty position for every collision

#### Disadvantage

Requires additional memory for the secondary data structure, such as linked lists.

#### 2 Linear Probing

Linear probing is an open-addressing technique.

When a collision occurs, the algorithm searches the next consecutive table locations until an empty location is found.

The probing function is:

h(k,i)=(h(k)+i)modm

where:

k = key
i = probe number
m = table size

For example, if the original index is 3:
 3 → 4 → 5 → 6 → ...
The algorithm checks each consecutive location until an empty position is found.

#### Advantage

Simple

Good cache performance


Does not require additional linked structures

#### Disadvantage

It suffers from primary clustering, where occupied locations form long groups.

### 3 Quadratic Probing

Quadratic probing is another open-addressing technique. Instead of checking consecutive positions, it uses a quadratic function to determine the next position.

The probing function is:

h(k,i)=(h(k)+i²)mod m

The probe positions are therefore determined by:

i=0,1,2,3,…

For example, if the initial hash value is 0:

0,1,4,9,…
#### Advantages
Reduces primary clustering

Spreads elements more widely than linear probing
#### Limitations
May not visit every table location

Requires careful selection of table size

Keys with the same initial hash value follow the same probe sequence

### 4 Double Hashing

Double hashing uses a second hash function to determine the probing interval.

The probing function is:

h(k,i)=(h1(k)+i×h2(k))modm

where:h1(k) = primary hash function
h2(k) = secondary hash function
i = probe number
m = table size

The first hash function determines the initial position, while the second hash function determines how far the algorithm moves after each collision.

##### Advantages

Reduces primary clustering

Reduces secondary clustering

Produces probe sequences that are closer to random probing

##### Limitations

Requires two hash calculations

More complex than linear probing

| Technique             | How collision is handled                              | Main advantage                           | Main disadvantage            |
| --------------------- | ----------------------------------------------------- | ---------------------------------------- | ---------------------------- |
| **Separate chaining** | Store colliding elements in a secondary structure     | Simple and flexible                      | Requires additional memory   |
| **Linear probing**    | Check consecutive locations                           | Simple and good cache performance        | Primary clustering           |
| **Quadratic probing** | Check locations using (i^2)                           | Reduces primary clustering               | May not visit every location |
| **Double hashing**    | Use a second hash function to determine the step size | Reduces primary and secondary clustering | More computationally complex |

### Question 4: Skip Lists

#### 4(a) Define a skip list

A skip list is a probabilistic data structure consisting of multiple levels of sorted linked lists.

The lowest level contains all elements, while the higher levels contain selected elements that act as shortcuts.

For example

Level 3: HEAD → 30 → 70

Level 2: HEAD → 10 → 30 → 50 → 70

Level 1: HEAD → 10 → 20 → 30 → 40 → 50 → 60 → 70 → 80

The bottom level contains all the elements.

The higher levels contain fewer elements and allow the search algorithm to skip over several nodes.


#### 4(b) Explain how a skip list improves search performance compared with a standard linked list

A standard linked list contains only one level.

For example:
    
  10 → 20 → 30 → 40 → 50 → 60 → 70 → 80

If we want to find 70, we may have to check:
    
    10 → 20 → 30 → 40 → 50 → 60 → 70
    
This requires sequential searching and has time complexity:

O(n)

    ##### Skip list

A skip list adds additional levels:

Level 3: HEAD → 30 → 70
                    ↓
Level 2: HEAD → 10 → 30 → 50 → 70
                              ↓
Level 1: HEAD → 10 → 20 → 30 → 40 → 50 → 60 → 70

    Instead of checking every element, the search starts at the highest level.

For example, to find 70:

Start at Level 3.
Move to 30.
Move to 70.
Find the target.

The upper levels therefore act as shortcuts.

| Structure            | Search method             | Expected search time |
| -------------------- | ------------------------- | -------------------: |
| Standard linked list | Sequential                |               (O(n)) |
| Skip list            | Multiple levels/shortcuts |          (O(\log n)) |

#### 4(c) Why is a skip list described as a probabilistic data structure?

A skip list is called probabilistic because the nodes that appear in the higher levels are selected using a random/probabilistic process.

In other words, we do not manually decide that every particular node must appear at Level 2 or Level 3.

Instead, the structure uses probability to determine how high a node is promoted.

Conceptually:

Level 3:          30              70

Level 2:      10  30      50      70

Level 1:      10  20  30  40  50  60  70  80

Some nodes appear at several levels, while others appear only at Level 1.

This probabilistic organization gives the skip list its expected:

O(logn)

#### 7(c) Explain how quadratic probing differs from linear probing
Linear probing

Linear probing checks consecutive locations:

h(k),h(k)+1,h(k)+2,…

For example:

0 → 1 → 2 → 3 → 4
Quadratic probing

Quadratic probing uses squared increments:

h(k),h(k)+1²,h(k)+2²,h(k)+3²,…

0 → 1 → 4 → 9 → 5

Therefore, quadratic probing spreads the probe positions more widely than linear probing and reduces primary clustering. However, it can have limitations such as not visiting every table location.

#### Question 9: Comparative Genomics Using Sets

In this question, Python sets are used to compare genes found in three cattle breeds: Ankole, N'Dama, and Boran.

Sets are useful because they contain unique elements and support operations such as:

Union

Intersection

Difference

```python
# Question 9: Comparative Genomics Using Sets

# Define the gene sets
ankole = {
    "EDN3",
    "PDE5A",
    "DNAH7",
    "ZAN",
    "RAB27B"
}

ndama = {
    "PDE5A",
    "DNAH7",
    "HSP90AA1",
    "RAB27B",
    "SLC24A5"
}

boran = {
    "DNAH7",
    "ZAN",
    "RAB27B",
    "HSP90AA1",
    "KIT"
}

# (a) All genes observed in at least one breed
all_genes = ankole | ndama | boran

# (b) Genes shared by all three breeds
shared_all = ankole & ndama & boran

# (c) Genes shared by Ankole and N'Dama
ankole_ndama = ankole & ndama

# (d) Genes found only in Ankole
only_ankole = ankole - ndama - boran

# (e) Genes found in Ankole but not in Boran
ankole_not_boran = ankole - boran

# (f) Genes found in exactly one breed
only_ndama = ndama - ankole - boran
only_boran = boran - ankole - ndama

exactly_one = only_ankole | only_ndama | only_boran

# (g) Number of unique genes in each breed
number_ankole = len(ankole)
number_ndama = len(ndama)
number_boran = len(boran)

# Display results
print("(a) All genes:", sorted(all_genes))
print("(b) Shared by all three:", sorted(shared_all))
print("(c) Shared by Ankole and N'Dama:", sorted(ankole_ndama))
print("(d) Only in Ankole:", sorted(only_ankole))
print("(e) Ankole but not Boran:", sorted(ankole_not_boran))
print("(f) Found in exactly one breed:", sorted(exactly_one))

print("(g) Number of unique genes:")
print("    Ankole:", number_ankole)
print("    N'Dama:", number_ndama)
print("    Boran:", number_boran)
```

### PART IV: SKIP LISTS
#### Question 11: Skip-List Search

A skip list is a probabilistic data structure made up of multiple levels of sorted linked lists. The lowest level contains all elements, while higher levels contain selected elements that act as shortcuts. This allows searching to skip over several elements and gives an expected search complexity of:

O(logn)

Given Skip List
Level 3:
HEAD ─────────────→ 30 ─────────────→ 70 → NULL

Level 2:
HEAD ─→ 10 ──────→ 30 ─→ 50 ──────→ 70 → NULL

Level 1:
HEAD → 10 → 20 → 30 → 40 → 50 → 60 → 70 → 80 → NULL


#### 11(a) Show the complete search path for key 60

We want to find:

60

The important rule is:

Start at the highest level, move right while the next key is smaller than the target, and move down when the next key would be greater than the target.

Step 1: Start at Level 3

Level 3 is:HEAD → 30 → 70 → NULL

We start at HEAD.

The next node is 30.

Since:

30<60

we move right to 30.

Now we are at:HEAD → 30

The next node is 70.

Since:

70>60

we cannot move to 70.

Therefore, we move down to Level 2.

Step 2: Move to Level 2

At Level 2, we are at node 30.

Level 2 is: HEAD → 10 → 30 → 50 → 70 → NULL

From 30, the next node is 50.

Since:

50<60

we move right to 50.

Now we are at 50.

The next node is 70.

Since:

70>60

we cannot move to 70.

Therefore, we move down to Level 1.

Step 3: Move to Level 1

At Level 1, we are at node 50.

Level 1 is: 10 → 20 → 30 → 40 → 50 → 60 → 70 → 80

The next node after 50 is 60.

Since:

60=60

we have found the key.

Therefore: 60 exists in the skip list
	
Complete search path for 60

The complete path is:

HEADL3→30L3↓30L2→50L2↓50L1→60L1

You can also show it visually:

                       │
                       ↓

                              │
                              ↓
Level 1:  HEAD → 10 → 20 → 30 → 40 → 50 → 60
                                                   ↑

                                                 FOUND


Final answer

Key 60 is found at Level 1.
