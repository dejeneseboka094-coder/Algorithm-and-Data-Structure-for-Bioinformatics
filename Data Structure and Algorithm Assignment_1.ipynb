### **Assignment 1: Dynamic Array / Array-Based List**

#Task: Implement your own dynamic array class in Python without using Python’s built-in list methods
such as append() for the main logic.

```python
class DynamicArray:

    def __init__(self):
        self.capacity = 2
        self.length = 0
        self.array = [None] * self.capacity

    def _resize(self, new_capacity):
        new_array = [None] * new_capacity

        for i in range(self.length):
            new_array[i] = self.array[i]

        self.array = new_array
        self.capacity = new_capacity

    def append(self, value):
        if self.length == self.capacity:
            self._resize(self.capacity * 2)

        self.array[self.length] = value
        self.length += 1

    def get(self, index):
        if index < 0 or index >= self.length:
            raise IndexError("Index out of bounds")
        return self.array[index]

    def set(self, index, value):
        if index < 0 or index >= self.length:
            raise IndexError("Index out of bounds")
        self.array[index] = value

    def insert(self, index, value):
        if index < 0 or index > self.length:
            raise IndexError("Index out of bounds")

        if self.length == self.capacity:
            self._resize(self.capacity * 2)

        for i in range(self.length, index, -1):
            self.array[i] = self.array[i - 1]

        self.array[index] = value
        self.length += 1

    def delete(self, index):
        if index < 0 or index >= self.length:
            raise IndexError("Index out of bounds")

        for i in range(index, self.length - 1):
            self.array[i] = self.array[i + 1]

        self.length -= 1
        self.array[self.length] = None

    def size(self):
        return self.length

    def display(self):
        return [self.array[i] for i in range(self.length)]



# 🧬 BIOINFORMATICS APPLICATION


dna = DynamicArray()

# Initial DNA sequence
initial_sequence = ["A", "T", "G", "C", "A", "G", "T"]

for base in initial_sequence:
    dna.append(base)

print("Initial DNA Sequence:", dna.display())



# 1. Access nucleotide at position 3

print("\nNucleotide at position 3:", dna.get(3))



# 2. Insert mutation at position 4

dna.insert(4, "T")
print("\nAfter insertion at position 4:", dna.display())



# 3. Delete nucleotide at position 2

dna.delete(2)
print("\nAfter deletion at position 2:", dna.display())



# 4. Replace nucleotide (set)

dna.set(1, "C")
print("\nAfter replacement at position 1:", dna.display())



# Final size check

print("\nFinal size of DNA sequence:", dna.size())
```

##### **Big-O Complexity Analysis**


| Operation | Complexity     | Reason                 |
| --------- | -------------- | ---------------------- |
| append    | O(1) amortized | occasional resize O(n) |
| get       | O(1)           | direct index access    |
| set       | O(1)           | direct overwrite       |
| insert    | O(n)           | shifting elements      |
| delete    | O(n)           | shifting elements      |
| resize    | O(n)           | copy all elements      |
| size      | O(1)           | stored variable        |

Dynamic arrays achieve O(1) amortized append by doubling capacity during resizing, trading occasional O(n) copy for overall efficiency.


### **Assignment 2: Single Linked List**

#Task: Implement a singly linked list.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:

    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)

        if not self.head:
            self.head = new_node
            return

        cur = self.head
        while cur.next:
            cur = cur.next
        cur.next = new_node

    def prepend(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def insert_after(self, target, data):
        cur = self.head

        while cur and cur.data != target:
            cur = cur.next

        if cur:
            new_node = Node(data)
            new_node.next = cur.next
            cur.next = new_node

    def delete(self, data):
        cur = self.head

        if cur and cur.data == data:
            self.head = cur.next
            return

        prev = None
        while cur and cur.data != data:
            prev = cur
            cur = cur.next

        if cur:
            prev.next = cur.next

    def search(self, data):
        cur = self.head

        while cur:
            if cur.data == data:
                return True
            cur = cur.next

        return False

    def display(self):
        cur = self.head
        result = []

        while cur:
            result.append(cur.data)
            cur = cur.next

        return " → ".join(result)



# 🧬 BIOINFORMATICS APPLICATION


assembly = SinglyLinkedList()

# Create genome assembly
assembly.append("Contig1")
assembly.append("Contig2")
assembly.append("Contig3")

print("Initial Assembly:")
print(assembly.display())

# Add new contig
assembly.append("Contig4")
print("\nAfter adding Contig4:")
print(assembly.display())

# Insert contig after Contig1
assembly.insert_after("Contig1", "Contig1.5")
print("\nAfter inserting Contig1.5:")
print(assembly.display())

# Remove low-quality contig
assembly.delete("Contig2")
print("\nAfter deleting Contig2:")
print(assembly.display())

# Search contigs
print("\nSearch Contig3:", assembly.search("Contig3"))
print("Search Contig2:", assembly.search("Contig2"))
```

##### **Big-O Complexity**

| Operation    | Complexity | Reason             |
| ------------ | ---------- | ------------------ |
| append       | O(n)       | traverse to end    |
| prepend      | O(1)       | direct head update |
| insert_after | O(n)       | search target      |
| delete       | O(n)       | search + relink    |
| search       | O(n)       | linear traversal   |
| display      | O(n)       | traverse all nodes |

**Singly linked lists** provide efficient insert/delete at known positions (O(1)), but require linear traversal (O(n)) for search and end operations, making them suitable for sequential biological data like contigs.


### **Assignment 3: Stack**

#Task Implement a stack using a Python class.

```python
class Stack:

    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return None

    def peek(self):
        if not self.is_empty():
            return self.items[-1]
        return None

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)



# 🧬 BIOINFORMATICS APPLICATION
# Sequence Alignment Traceback


traceback_stack = Stack()

# Push alignment operations
alignment_path = [
    "Match",
    "Match",
    "Insertion",
    "Match",
    "Deletion"
]

for op in alignment_path:
    traceback_stack.push(op)

print("Forward Traceback Path:")
print(traceback_stack.items)

# Reconstruct alignment in reverse (traceback)
reconstructed_path = []

while not traceback_stack.is_empty():
    reconstructed_path.append(traceback_stack.pop())

print("\nReconstructed Alignment Path (Reverse):")
print(reconstructed_path)
```

##### **Big-O Complexity**

| Operation | Complexity | Reason                    |
| --------- | ---------- | ------------------------- |
| push      | O(1)       | append at end             |
| pop       | O(1)       | remove last element       |
| peek      | O(1)       | direct access             |
| is_empty  | O(1)       | length check              |
| size      | O(1)       | stored length computation |

A stack follows  **LIFO (Last-In-First-Out)** behavior, making it ideal for sequence alignment traceback, where operations are reversed to reconstruct optimal alignment paths.


### **Assignment 4: Queue**

Task: Implement a queue.

```python
class Queue:

    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if not self.is_empty():
            return self.items.pop(0)
        return None

    def front(self):
        if not self.is_empty():
            return self.items[0]
        return None

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)



# 🧬 BIOINFORMATICS APPLICATION
# Sequencing Read Processing


reads_queue = Queue()

# Incoming sequencing reads
reads = ["Read1", "Read2", "Read3", "Read4"]

# Enqueue all reads
for r in reads:
    reads_queue.enqueue(r)

print("Initial Queue:")
print(reads_queue.items)

# Process reads in FIFO order
print("\nProcessing Order:")

while not reads_queue.is_empty():
    read = reads_queue.dequeue()
    print("Processing:", read)
```

##### **Big-O Complexity**

| Operation | Complexity | Reason                         |
| --------- | ---------- | ------------------------------ |
| enqueue   | O(1)       | append at end                  |
| dequeue   | O(n)       | shifting elements after pop(0) |
| front     | O(1)       | direct access                  |
| is_empty  | O(1)       | length check                   |
| size      | O(1)       | stored length                  |

A queue follows  **FIFO (First-In-First-Out)** behavior, making it ideal for sequencing read processing, where reads must be handled in the exact order they arrive.
